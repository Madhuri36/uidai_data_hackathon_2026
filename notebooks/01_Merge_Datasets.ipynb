# Merging All Subsets of the Datasets into 1 Full dataset and Saving it

```python
import pandas as pd
from glob import glob
```

```python
# Collect enrolment files
enrolment_files = sorted(glob("../data/raw/enrolment/*.csv"))
assert enrolment_files, "No enrolment files found!"

# Load and merge
dfs = []
for file in enrolment_files:
    df = pd.read_csv(file, low_memory=False)
    print(f"{file}: {df.shape}")
    dfs.append(df)

enrolment_df = pd.concat(dfs, ignore_index=True)

print("Merged enrolment shape:", enrolment_df.shape)
print("Columns:", enrolment_df.columns.tolist())

# Save merged dataset
enrolment_df.to_csv("../data/processed/enrolment_full.csv", index=False)
```
